```python
from google.colab import files
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

```python
uploaded = files.upload()
image_path = next(iter(uploaded.keys()))
```

```python
img = Image.open(image_path).convert('RGB')
img_array = np.array(img)

print("Ukuran gambar:", img.size)
print("Dimensi array:", img_array.shape)

plt.imshow(img)
plt.title("Citra Asli")
plt.axis('off')
plt.show()
```

```python
# Mengubah citra RGB menjadi grayscale
gray_array = np.dot(
    img_array[..., :3],
    [0.299, 0.587, 0.114]
).round().astype(np.uint8)

gray_img = Image.fromarray(gray_array, mode='L')

print("Grayscale berhasil dibuat!")
print("Ukuran gambar grayscale:", gray_img.size)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Citra Asli")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gray_img, cmap='gray')
plt.title("Hasil Grayscale")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
output_path = 'hasil_grayscale.png'

gray_img.save(output_path)

files.download(output_path)
```
